```python
# kiểm tra xem các cột dạng mã categorical có nằm trong tập giá trị được định nghĩa trong Data Dictionary của TLC hay không
valid_codes = {
    "VendorID": {1, 2, 6, 7},
    "RatecodeID": {1, 2, 3, 4, 5, 6, 99},
    "payment_type": {0, 1, 2, 3, 4, 5, 6},
    "store_and_fwd_flag": {"Y", "N"},
}

result = []

for col, valid in valid_codes.items():

    s = raw_data[col].dropna()

    invalid = s[~s.isin(valid)]

    result.append({
        "Column": col,
        "Invalid_count": len(invalid),
        "Invalid_pct": round(len(invalid) / len(raw_data) * 100, 4)
    })

# LocationID
for col in ["PULocationID", "DOLocationID"]:

    invalid = raw_data[
        (raw_data[col] < 1) |
        (raw_data[col] > 265)
    ]

    result.append({
        "Column": col,
        "Invalid_count": len(invalid),
        "Invalid_pct": round(len(invalid) / len(raw_data) * 100, 4)
    })

cat_result = pd.DataFrame(result)

cat_result
```

```python
RANGES = {
    "trip_distance":   (0,   100, False),   # > 0
    "fare_amount":     (0,   500, False),
    "total_amount":    (0,  1000, False),
    "passenger_count": (1,     6, True),    # >= 1
    "tip_amount":      (0,   200, True),    # >= 0
}


range_result = []

for c, (lo, hi, incl) in RANGES.items():

    s = raw_data[c]

    ok = s.notna()

    if incl:
        below = (ok & (s < lo)).sum()
    else:
        below = (ok & (s <= lo)).sum()

    above = (ok & (s > hi)).sum()

    range_result.append({
        "Column": c,
        "Null": int(s.isna().sum()),
        "Lower": int(below),
        "Upper": int(above),
        "Pct": round((below + above) / len(s) * 100, 3)
    })


range_result = pd.DataFrame(range_result)


print(range_result)
```

### Cross-column Validation

```python
dur2 = (
    raw_data.tpep_dropoff_datetime 
    - raw_data.tpep_pickup_datetime
).dt.total_seconds() / 60


sp2 = raw_data.trip_distance / (dur2 / 60).replace(0, np.nan)


CROSS = {
    "dropoff > pickup":      raw_data.tpep_dropoff_datetime > raw_data.tpep_pickup_datetime,
    "duration > 0 phút":     dur2 > 0,
    "duration <= 300 phút":  dur2 <= 300,
    "speed >= 0 mph":        sp2 >= 0,
    "speed < 80 mph":        sp2 < 80,
}


n = len(raw_data)

cross_result = []


for name, ok_mask in CROSS.items():

    invalid = int((~ok_mask.fillna(False)).sum())

    cross_result.append({
        "Rule": name,
        "Invalid_count": invalid,
        "Invalid_pct": round(invalid / n * 100, 3)
    })


cross_result = pd.DataFrame(cross_result)


print(cross_result)
```
